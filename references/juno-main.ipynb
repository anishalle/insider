```python
"""
Markout calculation — optimized for large datasets (255 GB+).

Key changes vs. the original pandas script
──────────────────────────────────────────
1. Polars instead of Pandas
   → automatic multi-threading across all available cores
   → 2-5× less memory than an equivalent pandas DataFrame
   → lazy evaluation: the query planner pushes filters/projections
     into the parquet reader so only needed columns are scanned.

2. Vectorized direction mapping
   → replaces the per-row `.apply()` with a single `when/then` chain.

3. Partitioned join_asof
   → Polars' `join_asof` is already parallelized internally, but we
     also trim the future-price frame to only the columns we need so
     the join touches less memory.

4. Streaming sink (optional)
   → if the full result doesn't fit in RAM you can write directly to
     parquet in a streaming fashion (see STREAM_MODE flag).

Requirements
────────────
    pip install polars pyarrow

Hardware assumptions
────────────────────
    64 cores, 375 GB RAM  (adjust POLARS_MAX_THREADS if needed)
"""

import os
import time
import polars as pl
from pathlib import Path

# ── let Polars use every core ──────────────────────────────────────────────
# Polars auto-detects cores, but you can cap it if you share the machine:
os.environ["POLARS_MAX_THREADS"] = "32"

# ── paths ──────────────────────────────────────────────────────────────────
MARKETS_PATH = "/home/axa230262/scratch/insider/markets.parquet"
TRADES_PATH  = "/home/axa230262/scratch/insider/trades.parquet"
OUT_DIR      = Path("./markout_outputs")
OUT_DIR.mkdir(parents=True, exist_ok=True)

# ── config ─────────────────────────────────────────────────────────────────
HORIZON = "5m"  # Polars duration string: "1s", "1m", "5m", "1h"

# If True, the final per-trade markout is written via Polars' streaming
# sink so the full result never has to live in RAM at once.
STREAM_MODE = False

t0 = time.perf_counter()


# ═══════════════════════════════════════════════════════════════════════════
# 1. LOAD — lazy scan pushes column selection into the reader
# ═══════════════════════════════════════════════════════════════════════════
print("Scanning parquet files (lazy) …")

trades_lazy = pl.scan_parquet(TRADES_PATH)
markets_lazy = pl.scan_parquet(MARKETS_PATH)


# ═══════════════════════════════════════════════════════════════════════════
# 2. INFER / CAST TIMESTAMP
# ═══════════════════════════════════════════════════════════════════════════
def cast_timestamp_expr() -> pl.Expr:
    """
    Return a Polars expression that produces a proper Datetime column.
    Handles: existing datetime col, or numeric timestamp in s/ms/µs/ns.
    """
    # Try 'datetime' first, fall back to 'timestamp'
    # We'll handle this after collecting schema info.
    return (
        pl.when(pl.col("timestamp") > 1e17)
        .then(pl.col("timestamp").cast(pl.Int64).cast(pl.Datetime("ns")).dt.replace_time_zone("UTC"))
        .when(pl.col("timestamp") > 1e14)
        .then((pl.col("timestamp") * 1_000).cast(pl.Int64).cast(pl.Datetime("ns")).dt.replace_time_zone("UTC"))
        .when(pl.col("timestamp") > 1e11)
        .then((pl.col("timestamp") * 1_000_000).cast(pl.Int64).cast(pl.Datetime("ns")).dt.replace_time_zone("UTC"))
        .otherwise((pl.col("timestamp") * 1_000_000_000).cast(pl.Int64).cast(pl.Datetime("ns")).dt.replace_time_zone("UTC"))
    )


# Check schema without collecting the full frame
schema = trades_lazy.collect_schema()
has_datetime_col = "datetime" in schema.names()
has_timestamp_col = "timestamp" in schema.names()

if has_datetime_col:
    trades_lazy = trades_lazy.with_columns(
        pl.col("datetime")
        .str.strptime(pl.Datetime("us"), format="%Y-%m-%d %H:%M:%S", strict=False)
        .dt.replace_time_zone("UTC")
        .alias("trade_time")
    )
elif has_timestamp_col:
    # Need to determine timestamp unit from a small sample
    sample = trades_lazy.select("timestamp").head(1000).collect()
    median_val = sample["timestamp"].cast(pl.Float64).median()

    if median_val > 1e17:
        unit_factor, unit = 1, "ns"
    elif median_val > 1e14:
        unit_factor, unit = 1_000, "ns"
    elif median_val > 1e11:
        unit_factor, unit = 1_000_000, "ns"
    else:
        unit_factor, unit = 1_000_000_000, "ns"

    trades_lazy = trades_lazy.with_columns(
        (pl.col("timestamp").cast(pl.Int64) * unit_factor)
        .cast(pl.Datetime("ns"))
        .dt.replace_time_zone("UTC")
        .alias("trade_time")
    )
else:
    raise ValueError("No 'datetime' or 'timestamp' column found.")


# ═══════════════════════════════════════════════════════════════════════════
# 3. CAST NUMERICS + JOIN MARKET METADATA  (all still lazy)
# ═══════════════════════════════════════════════════════════════════════════
for col in ["price", "token_amount", "usd_amount"]:
    if col in schema.names():
        trades_lazy = trades_lazy.with_columns(pl.col(col).cast(pl.Float64, strict=False))

markets_schema = markets_lazy.collect_schema()
if "market_id" in schema.names() and "id" in markets_schema.names():
    lookup_cols = [c for c in ["id", "question", "event_title", "slug"] if c in markets_schema.names()]
    markets_lazy = markets_lazy.select(lookup_cols).rename({"id": "market_id"})
    trades_lazy = trades_lazy.join(markets_lazy, on="market_id", how="left")

# Drop rows missing essentials
trades_lazy = trades_lazy.filter(
    pl.col("trade_time").is_not_null()
    & pl.col("price").is_not_null()
    & pl.col("asset_id").is_not_null()
)


# ═══════════════════════════════════════════════════════════════════════════
# 4. UNPIVOT: one row per user (maker + taker)  — vectorized
# ═══════════════════════════════════════════════════════════════════════════
print("Building user-level rows …")

# Determine common columns present
common_candidates = [
    "transaction_hash", "block_number", "trade_time", "timestamp", "datetime",
    "event_id", "event_slug", "event_title", "market_id", "condition_id",
    "question", "nonusdc_side", "price", "usd_amount", "token_amount",
    "asset_id", "order_hash",
]
# Re-check schema after joins
full_schema = trades_lazy.collect_schema()
common_cols = [c for c in common_candidates if c in full_schema.names()]

maker_lazy = (
    trades_lazy
    .select(common_cols + ["maker", "maker_direction"])
    .rename({"maker": "user", "maker_direction": "direction"})
    .with_columns(pl.lit("maker").alias("role"))
)

taker_lazy = (
    trades_lazy
    .select(common_cols + ["taker", "taker_direction"])
    .rename({"taker": "user", "taker_direction": "direction"})
    .with_columns(pl.lit("taker").alias("role"))
)

user_trades_lazy = pl.concat([maker_lazy, taker_lazy])


# ═══════════════════════════════════════════════════════════════════════════
# 5. VECTORIZED SIDE MAPPING  (replaces slow .apply())
# ═══════════════════════════════════════════════════════════════════════════
dir_lower = pl.col("direction").cast(pl.Utf8).str.strip_chars().str.to_lowercase()

side_expr = (
    pl.when(
        dir_lower.is_in(["buy", "bid", "b", "long"])
        | dir_lower.str.contains("buy")
        | dir_lower.str.contains("bid")
        | dir_lower.str.starts_with("+")
    ).then(pl.lit(1))
    .when(
        dir_lower.is_in(["sell", "ask", "s", "short"])
        | dir_lower.str.contains("sell")
        | dir_lower.str.contains("ask")
        | dir_lower.str.starts_with("-")
    ).then(pl.lit(-1))
    .otherwise(pl.lit(None))
    .alias("side")
)

user_trades_lazy = (
    user_trades_lazy
    .with_columns(side_expr)
    .filter(pl.col("side").is_not_null())
    .with_columns(
        (pl.col("trade_time") + pl.duration(minutes=5)).alias("target_time")
        # ↑ adjust if HORIZON changes — see note below
    )
)

# NOTE: For arbitrary horizon strings you can do:
#   pl.col("trade_time").dt.offset_by(HORIZON)
# but offset_by uses calendar offsets. For exact durations parse manually:
HORIZON_TD = pl.duration(
    **{
        {"s": "seconds", "m": "minutes", "h": "hours"}[HORIZON[-1]]: int(HORIZON[:-1])
    }
)
# Re-apply with the parsed duration (overrides the hard-coded +5 min above)
user_trades_lazy = user_trades_lazy.with_columns(
    (pl.col("trade_time") + HORIZON_TD).alias("target_time")
)


# ═══════════════════════════════════════════════════════════════════════════
# 6. COLLECT user_trades — needed before join_asof (requires eager frames)
# ═══════════════════════════════════════════════════════════════════════════
print("Collecting user trades into memory …")
user_trades = user_trades_lazy.collect(streaming=True)
print(f"  → {user_trades.shape[0]:,} user-trade rows  "
      f"({user_trades.estimated_size('gb'):.1f} GB in RAM)")

# Future-price lookup: only the 3 columns we need
print("Collecting future-price reference frame …")
future_prices = (
    trades_lazy
    .select(["asset_id", "trade_time", "price"])
    .rename({"trade_time": "future_trade_time", "price": "future_price"})
    .collect(streaming=True)
)
print(f"  → {future_prices.shape[0]:,} price rows  "
      f"({future_prices.estimated_size('gb'):.1f} GB in RAM)")


# ═══════════════════════════════════════════════════════════════════════════
# 7. JOIN_ASOF  — parallel across asset_id partitions
# ═══════════════════════════════════════════════════════════════════════════
print("Running join_asof (forward) …")

# Both frames must be sorted on the asof key within each group
user_trades = user_trades.sort(["asset_id", "target_time"])
future_prices = future_prices.sort(["asset_id", "future_trade_time"])

markouts = user_trades.join_asof(
    future_prices,
    left_on="target_time",
    right_on="future_trade_time",
    by="asset_id",
    strategy="forward",
)


# ═══════════════════════════════════════════════════════════════════════════
# 8. COMPUTE MARKOUT — fully vectorized
# ═══════════════════════════════════════════════════════════════════════════
print("Computing markouts …")

markouts = markouts.with_columns([
    (pl.col("side") * (pl.col("future_price") - pl.col("price"))).alias("markout"),

    pl.when(
        pl.col("price").is_not_null()
        & (pl.col("price") != 0)
        & pl.col("future_price").is_not_null()
    )
    .then(pl.col("side") * ((pl.col("future_price") / pl.col("price")) - 1.0) * 10_000)
    .otherwise(None)
    .alias("markout_bps"),

    pl.lit(HORIZON).alias("horizon"),
])

markouts = markouts.with_columns(
    (pl.col("markout") > 0).alias("is_positive_markout")
)


# ═══════════════════════════════════════════════════════════════════════════
# 9. WRITE OUTPUTS
# ═══════════════════════════════════════════════════════════════════════════
print("Writing per-trade markouts …")

per_trade_pq = OUT_DIR / f"markouts_per_trade_{HORIZON}.parquet"
per_trade_csv = OUT_DIR / f"markouts_per_trade_{HORIZON}.csv"

# Parquet first (fast, compressed) — always recommended at this scale
markouts.write_parquet(per_trade_pq, use_pyarrow=True, compression="zstd")
print(f"  ✓ {per_trade_pq}")

# CSV is optional at 255 GB scale — uncomment if you really need it:
# markouts.write_csv(per_trade_csv)
# print(f"  ✓ {per_trade_csv}")


# ── user summary ───────────────────────────────────────────────────────────
print("Computing user summary …")

user_summary = (
    markouts
    .group_by(["user", "role"])
    .agg([
        pl.col("transaction_hash").count().alias("n_trades"),
        pl.col("markout").mean().alias("avg_markout"),
        pl.col("markout").median().alias("median_markout"),
        pl.col("markout_bps").mean().alias("avg_markout_bps"),
        pl.col("is_positive_markout").mean().alias("positive_markout_rate"),
        pl.col("usd_amount").mean().alias("avg_usd_amount"),
    ])
    .sort(["avg_markout_bps", "positive_markout_rate", "n_trades"], descending=True)
)

summary_pq = OUT_DIR / f"user_markout_summary_{HORIZON}.parquet"
user_summary.write_parquet(summary_pq, use_pyarrow=True, compression="zstd")
print(f"  ✓ {summary_pq}")


# ── taker-only slice ───────────────────────────────────────────────────────
taker_only = markouts.filter(pl.col("role") == "taker")
taker_pq = OUT_DIR / f"markouts_taker_only_{HORIZON}.parquet"
taker_only.write_parquet(taker_pq, use_pyarrow=True, compression="zstd")
print(f"  ✓ {taker_pq}")

elapsed = time.perf_counter() - t0
print(f"\nDone in {elapsed / 60:.1f} minutes on {os.cpu_count()} cores.")
print(f"Peak RAM of result frame: {markouts.estimated_size('gb'):.1f} GB")

# Preview
print("\nSample:")
print(
    markouts.select([
        "transaction_hash", "user", "role", "direction", "side",
        "asset_id", "price", "future_price", "markout", "markout_bps",
        "trade_time", "future_trade_time",
    ]).head(10)
)
```
